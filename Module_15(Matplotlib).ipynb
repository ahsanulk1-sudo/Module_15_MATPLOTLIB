2D LINE PLOT


```python
import pandas as pd
import matplotlib.pyplot as plt

hours = [2,7,4,1,6,5,7,9]
days  =  [1,2,3,4,5,6,7,8]

plt.plot(days,hours)
```

```python
plt.xlabel("Days") 

plt.ylabel("Hours")

plt.title("Weekly Study Graph")

plt.plot(days,hours)
```

2D LINE PLOT FROM PANDAS DATA

```python
df = pd.read_csv('enrollment_data.csv') 
df
```

```python
plt.xlabel("Year") 

plt.ylabel("Count of Enrollment") 


plt.title("Student Enrollment Over Years")

plt.plot(df['Year'] ,df['Programming'])
```

```python
plt.xlabel("Year") 

plt.ylabel("Count of Enrollment") 


plt.title("Student Enrollment Over Years")

plt.plot(df['Year'] ,df['Programming'])

plt.plot(df['Year'] ,df['Digital Marketing'])

```

```python
#ADDING LABELS USING PLT.LEGEND()

plt.xlabel("Year") 

plt.ylabel("Count of Enrollment") 


plt.title("Student Enrollment Over Years")


plt.plot(df['Year'] ,df['Programming'],label = 'Programming')

plt.plot(df['Year'] ,df['Digital Marketing'], label = 'Digital Marketing')

plt.legend(loc = 'lower right')  # 'lower right will be in smaller letter
```

ADDING SOME VISUALIZATION

```python
plt.xlabel("Year") 

plt.ylabel("Count of Enrollment") 


plt.title("Student Enrollment Over Years")


plt.plot(df['Year'] ,df['Programming'],label = 'Programming',color = 'black',linewidth=3,linestyle = 'dashed',marker='o',markersize = 8)

plt.plot(df['Year'] ,df['Digital Marketing'], label = 'Digital Marketing',color ='red',linestyle = 'dotted')

plt.legend()
```

```python
#SEEING IN GRID

plt.xlabel("Year") 

plt.ylabel("Count of Enrollment") 


plt.title("Student Enrollment Over Years")


plt.plot(df['Year'] ,df['Programming'],label = 'Programming',color = 'black',linewidth=3,linestyle = 'dashed',marker='o',markersize = 8)

plt.plot(df['Year'] ,df['Digital Marketing'], label = 'Digital Marketing',color ='red',linestyle = 'dotted')

plt.legend()

plt.grid()

plt.show()  # USE IT EVERYTIME
```

SCATTER PLOT

```python
students = pd.read_csv('student_IQdata.csv') 
students
```

```python
#Relation between shoe size & Iq score
#NO CORRELATION

plt.xlabel('Shoe_Size')

plt.ylabel('IQ_Score')

plt.title('Relation between shoe size & Iq score') 

plt.scatter(students['Shoe_Size'] , students['IQ_Score'])
```

```python
#Relation between shoe size & Iq score
#POSITIVE CORRELATION

plt.xlabel('Study_Hour')

plt.ylabel('IQ_Score')

plt.title('Relation between study hour & Iq score') 

plt.scatter(students['Study_Hour'] , students['IQ_Score'])
```

```python
#Relation between Chiling_Hours & Iq score
#NEGATIVE CORRELATION

plt.xlabel('Chilling_Hours')

plt.ylabel('IQ_Score')

plt.title('Relation between Chilling_Hours & Iq score') 

plt.scatter(students['Chilling_Hours'],students['IQ_Score'],color=['green'])
```

BAR CHART

```python
import pandas as pd

data = {
    'Student': [ 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Male', 'Female', 'Male'],
    'StudyHours': [ 2, 5, 3, 7, 4, 8, 1],
    'Attendance': [60, 88, 70, 95, 80, 98, 55],
    'Grade': ['Fail', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail']
}

df = pd.DataFrame(data)
df
```

```python
#GENDERWISE 

gender_group = df.groupby('Gender').size()

gender_group.index    

gender_group.values
```

```python
plt.bar(gender_group.index,gender_group.values,color=['red','green'],edgecolor='black')
```

BAR CHART ON REAL DATA

```python
df = pd.read_csv('student_data.csv') 
df
```

```python
plt.xlabel('Completion Status')

plt.ylabel('Count')

plt.title('Bar Chart of Course Completion') 

status = df.groupby('CompletionStatus').size()

plt.bar(status.index,status.values,color=['red','green','blue'],edgecolor='black')
```

```python
#LOCATION

plt.xlabel('Location')

plt.ylabel('Count')

plt.title('Location of Students') 

status = df.groupby('Location').size()

plt.bar(status.index,status.values,color=['red','green','blue','orange','black'],edgecolor='black')

plt.show()
```

PIE CHART (CATEGORICAL & NUMEICAL)

```python


status = df.groupby('CompletionStatus').size()

plt.pie(status,labels = status.index,autopct='%1.1f%%',explode=(0.1,0.1,0.1),shadow=True)

plt.show()
```
